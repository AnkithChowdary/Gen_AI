### A function is a block of code that performs a specific task. Fucntions help in organizing code, reusing code and improving readability

```python
def even_or_code(num):
  if num%2==0:
    print("IT's a even")
  else:
    print("IT's a odd number")
```

```python
even_or_code(5)
```

```python
## function with multiple parameters
def add(a,b):
  return a+b
```

```python
## Default parameters
def greet(name="Guest"):
  return (f"Hello {name}!")
```

```python
print(greet("Ankit"))
print(greet())
```

```python
### varialbe length arguments
## Positional and keywords arguments

def print_numbers(*args):
  for arg in args:
    print(arg,end=",")
```

```python
print_numbers("ankith","Bhomraj",False,52)
```

```python
### Keyword arguments
## Here they are in the form of key value pairs
def print_details(**kwargs):
  for key,value in kwargs.items():
    print(f"{key}:{value}")
```

```python
print(print_details(name="Ankit",problems_solved="550",Language_used="C++"))
```

```python
### Keyword arguments
## Here they are in the form of key value pairs
def print_details(*args,**kwargs):

  
  for arg in args:
    print(f"Positional arguments: {arg}")

  for key,value in kwargs.items():
    print(f"{key}:{value}")
  
```

```python
print_details(4,1,name="Ankit",problems_solved="550",Language_used="C++")
```
