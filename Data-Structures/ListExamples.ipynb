## Real-World Examples using lists in python 
Example 1. Manage A to Do List

```python
to_do_list=["Buy Groceries","Clean the house","Solve two leetcode questions"]

## Adding new tasks
to_do_list.append("Solve problems in gfg")
to_do_list.append("Sleep for more than 7 hrs")

## Remove completed tasks
to_do_list.remove("Solve two leetcode questions")

## Checking if a task is in the list

if "Solve problems in gfg" in to_do_list:
  print("Make sure you solve!!!")

for task in to_do_list:
  print(task,end=" ")
```

## Example 2: Organizing Studnet Grades

Create a list to store and calculate average grades for students

```python
# Organizing student grades
grades=[69,92,88,73,99]

# adding a new grade
grades.append(95)

# Calculating the average grade
average_grade=sum(grades)/len(grades)
print(f"Average Grade: {average_grade:.2f}")

## Finding the highest and lowest grades
highest_grade=max(grades)
lowest_grade=min(grades)

print(f"The highest score is {highest_grade} and the lowest is {lowest_grade}")
```

## Inventory Management


```python
# Managing an inventory 
inventory =["apples","bananas","oranges","grapes"]

# Adding a new items
inventory.append("Strawberries")

# Removing an items that is out of stock
inventory.remove("bananas")

# Checking if an items is in stock
item="Oranges"

if item in inventory :
  print(f"{item} are in stock")
```
