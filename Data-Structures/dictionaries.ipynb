### Introduction to Dictionaries

Dictionaries are unordered collections of items. They Store the data in key-value pairs. Keys must be unique and immutable (eg strings,numbers, or tuples), while values can be of any type

```python
## Creating Dictionaries
empty_dict={}
print(type(empty_dict))
print(empty_dict)
```

```python
empty_dict=dict()
empty_dict
```

```python
student={"name":"Ankith","age":44,"grade":41}
print(student)
print(type(student))
print(student['name'])
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name',"Not available"))
```

```python
## Modifying Dictionary Elements
## Dictionary are mutable , so you can add, update, or delete elements
print(student)
student['age']=10 ## updating the value
print(student['age'])
student['last_name']="Doot" ## adding a new value
print(student)

del student['grade'] ## Delete key and value pair
print(student)
```

```python
## Dictionary methods

keys=student.keys() ## get all theh keys
print(keys)
values=student.values() ## get all values
print(values)

items=student.items() ## To get all key value pairs
print(items)
```

```python
## Shallow copy
studente_copy=student
print(student)
print(studente_copy)
```

```python
student['name']="Monika"
print(student)
print(studente_copy)
```

```python
student_copy1=student.copy() ## shallow copy
student['name']="Bhomraj"
print(student)
print(student_copy1)
```

```python
### Iterating over dictionaries 
## You can use loops to iterate over dictionaries ,key values , or items

## Iteratig over keys
for key in student.keys():
  print(key)
```

```python
## Iterate over values
for value in student.values():
  print(value)
```

```python
## Iterating over itmes
for key,val in student.items():
  print(key,val)
```

```python
## Nested Dictionaries

students={
  "student1":{"name":"Krish","age":32},
  "student2":{"name":"peter","age":10}
}

print(students)
print(students['student1']['age'])
print(students['student1'])
```

```python
## Dictionary comprehension
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## Condition dictionary comprehension
evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## practical Examples

## using a dictionary to count the frequence of elements in a list

lst=[1,2,1,1,4,2,3,5,6,4,4,4]
freq={}
for num in lst:
  if num in freq:
    freq[num]+=1
  else:
    freq[num]=1

print(freq)

freq2={}

for num in lst:
  freq2[num]=freq2.get(num,0)+1
print(freq2)
```
