## Tuples:
Tuples are ordered collections of items that are immutable. They are similar to lists

```python
## creating a tuple
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
numbers=tuple([1,2,3,4])
```

```python
mixed_tuple=("Ankith",5,True)
print(mixed_tuple)
```

```python
## Accessing tuple elements 

numbers
```

```python

print(numbers[2])
## similar to list
print(numbers[::2])

print(numbers[1::])
print(numbers[::-1])
```

```python
## Tuple operations
concatenated_tuple=numbers+mixed_tuple
concatenated_tuple
```

```python
## Entire tuple gets appended three times

mixed_tuple*3
```

```python
## Immutable nature of tuples
## Tuples are immutable, meaning thier elements cannot be changed once assigned

# lst=[1,2,3,4,5]
# print(lst)

# lst[1]="Krish"
# print(lst)

numbers[1]="Yogendra"

```

```python
## Tuple methods
print(numbers.count(1))
## gives the element index
print(numbers.index(4))
```

```python
## Packing and unpacking of tuples

pack_tuple="monika",5,True
pack_tuple
```

```python
## Unpacking tuple
a,b,c=pack_tuple
print(c)
```

```python
## Unpacking with *
numbers=(1,2,4,5,6)
first,*middle,last=numbers
print(first)
## It gives the elements expect first element upto the end except the last element
print(middle)
print(last)
```

```python
## Nested Tuple

lst=[[1,2,3,4],[6,7,8,9],[1,"Hello",3.14,"c"]]

print(lst[0][0:3])
```

```python
## Nested Tuple

lst=[[1,2,3,4],[6,7,8,9],(1,"Hello",3.14,"c")]

print(lst[0][0:3])
```

```python
nested_tuple=((1,2,3),("a","b","c"),(True,False))
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## Iterating over nested tuples

for sub in nested_tuple:
  for item in sub:
    print(item,end=" ")
```
