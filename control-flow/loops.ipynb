```python
# While loop

count=0

while count < 5:
  print(count)
  count=count+1

```

```python
## break statement

for i in range(10):
  if i==5:
    break
  print(i)
```

```python
## continue

for i in range(10):
  if i%2==0:
    continue
  print(i)
```

```python
## pass
## The pass statement is a null operation  it does nothing

for i in range(5):
  if i==3:
    pass
  print(i)
```

```python
## Nested loops
## a loop inside a loop

for i in range(3):
  for j in range(2):
    print(f"i:{i} and j:{j}")
```

```python
## Examples: calculate the sum of n natural number using while and for loop

# while loop

n=10
sum=0
count=1

while count<=n:
  sum=sum+count
  count=count+1

print("Sum of first 10 natural number is ",sum)
```

```python
n=10
sum=0
for i in range(n+1):
  sum=sum+i
print(sum)
```

```python
## Prime number between 1 and 10

for i in range(2,100):
  for j in range(2,int(i/2+1)):
    if(i%j==0):
      break
  else:
    print(i, " is a prime number")
```
